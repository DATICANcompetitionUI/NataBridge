# Exploratory Data Analysis (EDA)

#### Loading the data

```python
import pandas as pd

df = pd.read_csv('../data/raw/uci_dataset_train.csv')
df.head()
```

#### Data type of values per column

```python
df.info()
```

#### Dataset Overview

```python
num_of_records, num_of_columns = df.shape

print(f"Total number of records in the dataset: {num_of_records}")
print(f"Number of columns: {num_of_columns}")

features_list = df.columns.tolist()[:-1]
print(f"\nFeatures examined: {features_list}")
print(f"Number of features examined: {num_of_columns - 1}")
```

#### Dataset Description - Missing Values

```python
print("NOTE: Each column is paired with the number of missing values in them")

df.isnull().sum()
```

#### Dataset Description - Duplicates

```python
print(f"\nAny duplicate: {bool(df.duplicated().any())}")
print(f"Number of duplicate records: {int(df.duplicated().sum())}")
```

#### Class Balance

```python
import seaborn as sns

risk_level_stats = df["RiskLevel"].value_counts().to_dict()
print(risk_level_stats)


sns.countplot(data=df, x='RiskLevel')
```

#### Feature Distribution

```python
sns.histplot(df['Age'])
```

```python
sns.histplot(df['BS'])
```

```python
sns.histplot(df['BodyTemp'])
```

```python
sns.histplot(df['SystolicBP'])
```

```python
sns.histplot(df['DiastolicBP'])
```

#### Outliers

```python
sns.boxplot(x=df['BS'])
```

#### Correlation Analysis

```python
corr = df.corr(numeric_only=True)

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm')
```

#### Relationship Between Features and Risk

```python
sns.boxplot(
    x='RiskLevel',
    y='Age',
    data=df
)
```
