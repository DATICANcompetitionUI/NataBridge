# Result Explanation

#### Load the model

```python
import joblib
import pandas as pd

xgboost = joblib.load("../models/xgboost.pkl")

x_test = pd.read_csv("../data/processed/test/x_test.csv")
```

#### Creating the explainer

```python
import shap
import matplotlib.pyplot as plt

explainer = shap.TreeExplainer(xgboost)
```

#### SHAP values

```python
shap_values = explainer(x_test)
```

#### Global Explanation

```python
# this gives explanation for the high risk (representing 2 in the encoding) in the test set
shap.plots.beeswarm(shap_values[:, :, 2])
```

```python
# this gives explanation for the mid risk (representing 1 in the encoding) in the test set
shap.plots.beeswarm(shap_values[:, :, 1])
```

```python
# this gives explanation for the low risk (representing 0 in the encoding) in the test set
shap.plots.beeswarm(shap_values[:, :, 0])
```

#### Explanation for a single person

```python
# explain the first patient's prediction for High Risk (class 2)
shap.plots.waterfall(shap_values[0, :, 2])
```
