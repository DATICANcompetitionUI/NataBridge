# Model Training

#### Load the training dataset

```python
import pandas as pd

x_train = pd.read_csv("../data/processed/train/x_train.csv").squeeze()
y_train = pd.read_csv("../data/processed/train/y_train.csv").squeeze()
```

#### Train the model

```python
from sklearn.ensemble import RandomForestClassifier

random_forest = RandomForestClassifier(random_state=30, n_estimators=200, max_depth=15, min_samples_split=2, min_samples_leaf=1)

random_forest.fit(x_train, y_train)
```

```python
from xgboost import XGBClassifier

# XGBoost
xgboost = XGBClassifier(random_state=30, learning_rate=0.1, max_depth=7, n_estimators=100, subsample=0.8, colsample_bytree=0.8)
xgboost.fit(x_train, y_train)
```

#### Hyperparameters Tuning and Cross-validation

```python
from sklearn.model_selection import cross_val_score

# scores = cross_val_score(random_forest, x_train, y_train, cv=5, scoring="accuracy")
scores = cross_val_score(xgboost, x_train, y_train, cv=5, scoring="accuracy")

print(scores.mean())
```

#### Save the tuned model

```python
import joblib

# joblib.dump(random_forest, "../models/random_forest.pkl")
joblib.dump(xgboost, "../models/xgboost.pkl")
```
