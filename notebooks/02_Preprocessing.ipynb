# Preprocessing

### loading of the dataset

```python
import pandas as pd

df = pd.read_csv('../data/raw/uci_dataset_train.csv')
```

#### separation of features from target

```python
# x holds only features
x = df.drop("RiskLevel", axis=1)

# y holds only the target
y = df["RiskLevel"]
```

#### encoding of the target

```python
mapping = {
    "low risk":0,
    "mid risk":1,
    "high risk":2
}

y = y.map(mapping)
```

#### train test split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=30, stratify=y)
```

#### feature Scaling

Feature scaling was not applied because the selected models (Random Forest and XGBoost) are threshold-based tree algorithms. These models make predictions by comparing feature values against learned thresholds rather than measuring distances between samples. Consequently, scaling numerical features does not significantly affect their performance.

#### verification of train test split

```python
train_num_of_records, num_of_features = x_train.shape
test_num_of_records, num_of_features = x_test.shape

print(f"Number of features: {num_of_features}")
print(f"Number of 'train' records in the dataset: {train_num_of_records}")
print(f"Number of 'test' records in the dataset: {test_num_of_records}")
print(f"Total records preprocessed: {train_num_of_records} + {test_num_of_records} = {train_num_of_records + test_num_of_records}")
```

```python
print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```
